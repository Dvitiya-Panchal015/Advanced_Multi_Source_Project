```python
Top_Market_Cap = df.nlargest(10, "Market_Cap")

Top_Market_Cap[
    ["Name", "Market_Cap", "Market_Cap_Rank"]
]
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="Name",
    y="Market_Cap",
    color = "#C94845"
)

plt.title("Top 10 Cryptocurrencies by Market Capitalization")
plt.xlabel("Market Capitalization")
plt.ylabel("Cryptocurrency")

plt.show()
```

```python
Top_Volume = df.nlargest(10, "Total_Volume")

Top_Volume[
    ["Name", "Total_Volume"]
]
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="Name",
    y="Total_Volume",
    color = "Pink"
)

plt.title("Top 10 Cryptocurrencies by 24-Hour Trading Volume")
plt.xlabel("Trading Volume")
plt.ylabel("Cryptocurrency")

plt.show()
```
