# Advanced EDA

```python
# Outlier Analysis and Impact (Box plot)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(data=df, y="Current_Price", ax=axes[0])
axes[0].set_title("Current Price - Outliers")

sns.boxplot(data=df, y="Market_Cap", ax=axes[1])
axes[1].set_title("Market Cap - Outliers")

sns.boxplot(data=df, y="Total_Volume", ax=axes[2])
axes[2].set_title("Trading Volume - Outliers")

plt.tight_layout()
plt.show()
```

```python
# Impact of outliers

# Impact of Outliers - Current Price

# Calculate Q1 and Q3
Q1 = df["Current_Price"].quantile(0.25)
Q3 = df["Current_Price"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
price_outliers = df[
    (df["Current_Price"] < lower_limit) |
    (df["Current_Price"] > upper_limit)
]

# Data without outliers
price_without_outliers = df[
    (df["Current_Price"] >= lower_limit) &
    (df["Current_Price"] <= upper_limit)
]

# Results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nNumber of Outliers:", len(price_outliers))

print("\nMean Price - All Data:",
      df["Current_Price"].mean())

print("Median Price - All Data:",
      df["Current_Price"].median())

print("\nMean Price - Without Outliers:",
      price_without_outliers["Current_Price"].mean())

print("Median Price - Without Outliers:",
      price_without_outliers["Current_Price"].median())
```

```python
# # Statistical tests (t-test / ANOVA if needed)

df["Market_Cap_Category"] = pd.qcut(
    df["Market_Cap"],
    q=3,
    labels=["Low", "Medium", "High"]
)
```

```python
from scipy.stats import f_oneway
```

```python
# ANOVA

low = df[
    df["Market_Cap_Category"] == "Low"
]["Price_Change_Percentage_24H"].dropna()

medium = df[
    df["Market_Cap_Category"] == "Medium"
]["Price_Change_Percentage_24H"].dropna()

high = df[
    df["Market_Cap_Category"] == "High"
]["Price_Change_Percentage_24H"].dropna()

f_stat, p_value = f_oneway(
    low,
    medium,
    high
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
# Interpretation

if p_value < 0.05:
    print(
        "Result: There is a statistically significant "
        "difference in price change among the market-cap groups."
    )
else:
    print(
        "Result: There is no statistically significant "
        "difference in price change among the market-cap groups."
    )
```
