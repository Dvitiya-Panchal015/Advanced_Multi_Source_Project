```python
# clean the price column
df["price"] = (
    df["price"]
    .astype(str)
    .str.replace("Â", "", regex=False)
    .str.replace("£", "", regex=False)
    .astype(float)
)
```

```python
# All column names in Title Case

df.rename(columns={"title":"Title",
                   "price":"Price",
                   "rating":"Rating",
                   "instock":"In_stock",
                   "category":"Category",
                   },inplace = True)
```

```python
df
```

```python
rating_tag = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating"] = df["Rating"].map(rating_tag)
```

```python
# clean the column availability
df["In_stock"] = df["In_stock"].str.strip()
```

```python
# clean the column category
df["Category"] = df["Category"].str.strip()
```

```python
# cleaning the title columns
df["Title"] = df["Title"].str.strip()
```
